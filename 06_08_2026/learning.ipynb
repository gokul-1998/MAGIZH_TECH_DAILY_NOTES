```python
import time

def make_coffee():
    print("Making coffee...")
    time.sleep(3)
    print("Coffee Ready")

def make_sandwich():
    print("Making sandwich...")
    time.sleep(2)
    print("Sandwich Ready")

make_coffee()
make_sandwich()

print("Breakfast Ready!")
```

```python
import asyncio
import time

async def make_coffee():
    print("Making coffee...")
    await asyncio.sleep(3)
    print("Coffee Ready")

async def make_sandwich():
    print("Making sandwich...")
    await asyncio.sleep(2)
    print("Sandwich Ready")

async def main():
    await asyncio.gather(
        make_coffee(),
        make_sandwich()
    )

    print("Breakfast Ready!")

await main()
```
