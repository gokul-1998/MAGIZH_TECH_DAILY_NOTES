3) what is venv , and why do we need venv
    - venv is used to handle dependencies like libraries and packages

    - We need venv in Python to `isolate project dependencies and prevent version conflicts between different applications`.
    - By default, Python installs all packages into a single, global directory. If two projects require different versions of the same library, they will overwrite each other and cause system-critical errors.
    - Core Reasons to Use venvPrevents Dependency Conflicts: Different projects can use different versions of the same package (e.g., Project A uses Django 3.2 while Project B uses Django 5.0) without interference.
    - Keeps System Clean: It avoids cluttering your global Python installation with hundreds of temporary packages that you only need for a single project.
    - Ensures Reproducibility: It allows you to generate a precise list of dependencies (usually via requirements.txt), making it easy for other developers to replicate your exact environment.

5) what is microservice architecture and monolithic architecture?
    - Monolithic architecture is a `traditional model where all components of an application are combined into a single program`. In contrast, microservices architecture breaks down the application into smaller, independent services that communicate with each other, allowing for greater flexibility, scalability, and maintainability.
    - microservice architecture is a design approach where an application is built as a collection of small, loosely coupled services, each responsible for a specific functionality. These services communicate with each other through APIs, allowing for independent development, deployment, and scaling. In contrast, monolithic architecture is a traditional approach where all components of an application are tightly integrated into a single codebase, making it harder to manage and scale as the application grows.
